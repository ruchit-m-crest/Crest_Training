## 1.Find the indices of non-zero elements in a list.

```python
lst = [1, 2, 0, 0, 4, 0]
non_zero_indices = [i for i, x in enumerate(lst) if x != 0]
print(non_zero_indices)
```

## 2.Normalize the values in a list between 0 and 1.

```python
lst = [2, 4, 16, 1, 7]
min_val, max_val = min(lst), max(lst)
normalized_lst = [(x - min_val) / (max_val - min_val) for x in lst]
print(normalized_lst)
```

## 3.Check if a number is prime.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
print(is_prime(29))
print(is_prime(18))
```
