## 1. Find the pair  of  maximum and minimum product from a given list

```python
import  itertools as it

def list_min_max(nums):

    result_max = max(it.combinations(nums, 2), key = lambda sub: sub[0] * sub[1])
    result_min = min(it.combinations(nums, 2), key = lambda sub: sub[0] * sub[1])

    return result_max, result_min

nums = [2,6,7,8,9,3,6,5,10,4]

print('Pairs of maximum and minimum product from the list:')

print(list_min_max(nums))
    
```
