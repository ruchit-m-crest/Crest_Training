## 1. Program to filter cities by population

```python
cities = [ 
    ("delhi", 30000000),
    ("Mumbai", 20000000),
    ("Surat", 10700000),
    ("Ahmedabad", 10000000),
    ("Rajkot", 1600000),
    ("Kolkata", 1500000),
    ("Jaipur", 1500000),
]

print('List of cities and population : ')
print(cities)

def population(cities):
    return cities[1] > 10000000

million_plus_cities = list(filter(population, cities))

print('List of  cities over 10-million population')
print(million_plus_cities)
    
```
