## 1.Handling FileNotFoundError in python file handling program

```python
def open_file(Filename):
    try:
        file = open(filename,'r')

        contents = file.read()

        print('File Contents')

        print(contents)

        file.close()
        
    except FileNotFoundError:
        print('Errror : File not Found')

filename = input('Enter a file name :')

open_file(filename)
    
```
