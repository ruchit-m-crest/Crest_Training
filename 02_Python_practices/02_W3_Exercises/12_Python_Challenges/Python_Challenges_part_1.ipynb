## 1. Python program to reverse the digit of an integer

```python
def reverse_integer(x):
    sign = -1 if x < 0 else 1
    x *= sign  

    while x:
        if x % 10 == 0:
            x //= 10 
        else:
            break

    x = str(x)
    lst = list(x)  
    lst.reverse()
    x = ''.join(lst)
    x = int(x)

    return sign * x

print(reverse_integer(234))
print(reverse_integer(-234))
```

## 2. Compute the sum of digits of the number 2^20

```python
def digit_sum():
    n = 2**20
    ans = sum(int(c) for c in str(n))
    return str(ans)

print(digit_sum())
```

## 3. Takes a string of numbers and letters and return string which consists of letters

```python
def decode(str1):
    ints = "1234567890"

    num = ""
    letters = ""
    result_string = ""

    i = 0

    while i < len(str1):
    
        if str1[i] in ints:
            num += str1[i]
              
        else:
            letters += str1[i]

        i += 1

    for i, char in enumerate(num):

        result_string += int(char) * letters[i]

    return result_string
    
print(decode("4A3B2C1D2A"))
```
