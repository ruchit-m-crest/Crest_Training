## 1.Display Enum member names ordered by their values

```python
import enum
class Country(enum.IntEnum):
    Afghanistan=93
    India = 91
    Usa = 1
    Angola = 244
    Antractica= 672
    Albania = 355

print('Country name orderded by country code')
print('\n'.join(' ' + c.name for c in sorted(Country)))
```
