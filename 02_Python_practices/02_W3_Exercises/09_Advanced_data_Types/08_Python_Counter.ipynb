## 1. Sum of Two Counter

```python
from collections import Counter
```

```python
def add_counter(ctr1, ctr2):
    return ctr1 + ctr2

def main():
    counter1 = Counter({'item1':10, 'item2':20, 'item3':30})
    counter2 = Counter({'item1':5, 'item2':15, 'item3':25})

    result_counter = add_counter(counter1, counter2)

    print('Counter 1:', counter1)
    print('Counter 2:', counter2)
    print('sum counter:', result_counter)

if __name__ == '__main__':
    main()
    
```
