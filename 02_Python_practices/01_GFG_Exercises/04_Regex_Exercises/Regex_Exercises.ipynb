## 1.program to find numbers from a string

import re

def number_matcher(text):
    
    pattern = r'\d+'
    matches = re.findall(pattern, text)
    
    if matches:
        print(" ".join(matches))
    else:
        print(-1)


test = 'asdasd123asmdasdk34234kfdsd34sdfk5'
number_matcher(test)


## 2.Program to validate the password

```python
def validate(str):
    pat= '[a-z]+[!@#$%]+[0-9]'
    
    match = re.search(pat, str)
    if(match):
        return True
    else:
        return False

str = 'asdsab@!@234'
validate(str)
```

## 3.Program to extract Strings between HTML Tags

```python
import re

test_str = '<b>Gfg</b> is <b>Best</b>. I love <b>Reading CS</b> from it.'
print("The original string is : " + test_str)

tag = "b"

# Regex pattern to extract content inside <b>...</b>
reg_str = "<" + tag + ">(.*?)</" + tag + ">"
res = re.findall(reg_str, test_str)

print("The Strings extracted : " +  " ".join(res))
```
