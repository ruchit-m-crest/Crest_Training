## 1. Remove loop in linked list

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.next = None

class Solution:
    def removeLoop(self, head):
        if head is None:
            return

        visited = set()
        prev = None
        curr = head

        while curr:
            if curr in visited:
                prev.next=None
                break
            visited.add(curr)
            prev = curr
            curr = curr.next


def createLinkedList(values):
    if not values:
        return None

    head = Node(values[0])
    current = head
    for val in values[1:]:
        current.next = Node(val)
        current = current.next

    return head


def createLoop(head, pos):
    if pos == -1:
        return

    loop_start = head
    for _ in range(pos):
        loop_start = loop_start.next

    tail = head
    while tail.next:
        tail = tail.next

    tail.next = loop_start  # Creates the loop



def printLinkedList(head, limit=50):
    current = head
    count = 0
    while current and count < limit:
        print(current.data, end=" -> ")
        current = current.next
        count += 1

    print("None" if count < limit else "... (loop detected or long list)")


if __name__ == "__main__":
    # Example: Linked list = 1 -> 2 -> 3 -> 4 -> 5, with loop: 5 -> 2
    values = [1, 2, 3, 4, 5]
    head = createLinkedList(values)

    createLoop(head, 1)  # Creates a loop at position 1 (2nd node)

    print("Before removing loop:")
    printLinkedList(head)  # Will stop after 50 nodes if loop exists

    # Remove the loop
    sol = Solution()
    sol.removeLoop(head)

    print("\nAfter removing loop:")
    printLinkedList(head)  # Should end with None
```

## 2.python program to search element in a circular linked list

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def add(self, data):
        newNode = Node(data)

        if self.head is None:
            self.head = newNode
            self.tail = newNode
            newNode.next = self.head  # Point to itself for circularity
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head  # Maintain circular link

    def findNode(self, element):
        if self.head is None:
            print("Empty list")
            return

        current = self.head
        found = False

        while True:
            if current.data == element:
                found = True
                break
            current = current.next
            if current == self.head:
                break

        if found:
            print("Element is present")
        else:
            print("Element is not present")


if __name__ == '__main__':
    circularLinkedList = CircularLinkedList()

    circularLinkedList.add(1)
    circularLinkedList.add(2)
    circularLinkedList.add(3)
    circularLinkedList.add(4)
    circularLinkedList.add(5)
    circularLinkedList.add(6)

    circularLinkedList.findNode(2)
    circularLinkedList.findNode(7)
```

## 3.Stack using Doubly linked list

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class Stack:
    def __init__(self):
        self.head = None

    def push(self, data):
        if self.head is None:
            self.head = Node(data)
        else:
            new_node = Node(data)
            self.head.prev = new_node
            new_node.next = self.head
            self.head = new_node

    def pop(self):
        if self.head is None:
            return None
        elif self.head.next is None:
            temp = self.head.data
            self.head = None
            return temp
        else:
            temp = self.head.data
            self.head = self.head.next
            self.head.prev = None
            return temp

    def top(self):
        return self.head.data

    def size(self):
        temp = self.head
        count = 0
        while temp is not None:
            count += 1
            temp = temp.next
        return count

    def isEmpty(self):
        return self.head is None

    def printstack(self):
        print("stack elements are:")
        temp = self.head
        while temp is not None:
            print(temp.data, end="->")
            temp = temp.next

if __name__ == '__main__':
    stack = Stack()
    print("Stack operations using Doubly LinkedList")
    stack.push(4)
    stack.push(5)
    stack.push(6)
    stack.push(7)
    stack.printstack()
    print("\nTop element is", stack.top())
    print("Size of the stack is", stack.size())
    stack.pop()
    stack.pop()
    stack.printstack()
    print("\nstack is empty:", stack.isEmpty())
```

## 4.Python program to find middle of a linked list using one traversal

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class NodeOperation:
    def pushNode(self, head_ref, data_val):
        new_node = Node(data_val)
        new_node.next = head_ref
        head_ref = new_node
        return head_ref

    def printNode(self, head):
        while head is not None:
            print('%d->' % head.data, end="")
            head = head.next
        print("NULL")

    def getLen(self, head):
        temp = head
        length = 0
        while temp is not None:
            length += 1
            temp = temp.next
        return length

    def printMiddle(self, head):
        if head is not None:
            length = self.getLen(head)
            temp = head
            midIdx = length // 2
            while midIdx != 0:
                temp = temp.next
                midIdx -= 1
            print('The middle element is:', temp.data)

head = None
temp = NodeOperation()
head = temp.pushNode(head, 5)
head = temp.pushNode(head, 4)
head = temp.pushNode(head, 3)
head = temp.pushNode(head, 2)
head = temp.pushNode(head, 1)
temp.printNode(head)
temp.printMiddle(head)
```
